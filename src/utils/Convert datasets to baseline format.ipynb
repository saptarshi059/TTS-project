```python
from datasets import load_dataset
from pathlib import Path
import json
```

# Agent-Distillation

## 2wiki/hotpotqa/musique

```python
dataset_mapping = {"2wikimultihopqa": "2wikimultihopqa_500_20250511", 
                   "hotpotqa": "hotpotqa_500_20250422", 
                   "musique": "musique_500_20250504"
                  }

new_metadata = {'dataset_info': {'name': 'blank',
                                 'fold': 'test', 
                                 'examples_range': {'start': 0, 
                                                    'end': 1000},
                                 'total_examples': 1000,
                                 'creation_date': '2025-05-11',
                                 'filtering_criteria': None}}

for dataset in dataset_mapping:
    with Path(f"../../data/{dataset}/test.json").open("r") as file:
        my_version = json.load(file)
        
    new_metadata["dataset_info"]["name"] = dataset
    formatted_samples = []
    for sample in my_version:
        formatted_samples.append({'id': sample["id"],
                                  'question': sample["question"],
                                  'answer': sample["answer"],
                                  'level': 'hard',
                                  'type': 'compositional',
                                  'dataset_name': dataset,
                                  'split': 'test'})

    processed_dict = {"metadata": new_metadata,
                     "examples": formatted_samples}
    
    with Path(f"../baselines/agent-distillation/data_processor/qa_dataset/test/{dataset_mapping[dataset]}.json").open("w") as file:
        json.dump(processed_dict, file, indent=4)
```

## frames

```python
frames = load_dataset("google/frames-benchmark")['test']

new_metadata = {'dataset_info': {'name': 'frames',
                                 'fold': 'test', 
                                 'examples_range': {'start': 0, 
                                                    'end': 1000},
                                 'total_examples': 824,
                                 'creation_date': '2025-05-11',
                                 'filtering_criteria': None}}

formatted_samples = []
for idx, row in enumerate(frames):
    formatted_samples.append({'id': str(idx),
                              'question': row["Prompt"],
                              'answer': row["Answer"],
                              'level': 'hard',
                              'type': 'compositional',
                              'dataset_name': 'frames',
                              'split': 'test'})

processed_dict = {"metadata": new_metadata, "examples": formatted_samples}

with Path("../baselines/agent-distillation/data_processor/qa_dataset/test/frames.json").open("w") as file:
    json.dump(processed_dict, file, indent=4)
```
